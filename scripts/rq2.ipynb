# Import Dependencies

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Dataset Info

```python
df = pd.read_csv(r"dataset\master_student_dataset.csv")
df.info()
```

```python
df['source'].value_counts()
```

```python
sns.histplot(df['academic_score'], kde=True)
```

```python
sns.heatmap(df.select_dtypes('number').corr(), annot=True, cmap='coolwarm')
```

```python
df_habits = df[df['source'] == 'habits'].drop(columns=['source'])
print(df_habits.shape)  # should be (1000, 8)

# Re-run your heatmap
sns.heatmap(df_habits.select_dtypes('number').corr(), 
            annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap - Habits Source Only')
plt.show()
```

```python
df_habits = df[df['source'] == 'productivity'].drop(columns=['source'])
print(df_habits.shape)  # should be (1000, 8)

# Re-run your heatmap
sns.heatmap(df_habits.select_dtypes('number').corr(), 
            annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap - Productivity Source Only')
plt.show()
```
